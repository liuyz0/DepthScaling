```python
import torch
import matplotlib.pyplot as plt
```

```python
models = ['70m', '160m', '1b', '1.4b', '2.8b', '12b']
```

```python
data = [torch.load(f'../outputs/pythia-{model}.pt') for model in models]
```

```python
data[0]['tokens']
```

```python
print([data[i]['p_clean'] for i in range(len(models))])
```

```python
print([data[i]['p_corrupt'] for i in range(len(models))])
```

```python
for i in [5]:
    plt.figure(figsize=(3.4, 2.1), constrained_layout=True)
    plt.imshow(data[i]['results'][:,:-1].T, cmap='Greys', 
               extent=(0.5 / data[i]['results'].shape[0], (data[i]['results'].shape[0]+0.5) / data[i]['results'].shape[0], 5.5, -0.5), aspect='auto')
    plt.yticks(range(6), [data[i]['tokens'][j] for j in range(6)], rotation=30)
    print(data[i]['results'].shape[0])
    plt.xlabel(r'Normalized index, $l/\ell$')
    plt.colorbar()
    plt.savefig(f'../figures/causal_trace.pdf')
    plt.show()
```
